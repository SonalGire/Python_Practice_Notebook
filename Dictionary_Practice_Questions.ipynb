57. Search key using loop.

```python
a = {'name': 'Alice', 'age': 30, 'city': 'New York', 'height_cm': 170.5, 'zip_code': '10001'}

search_key = input("Enter a key to search:")
found =  False

for key in a.keys():
  if key == search_key:
    print(f"{search_key} is present")
    found = True
    break
if not found:
  print(f"{search_key} is not present")
```
