8.Write a program to find the second largest number in a list.

```python
a = [1,3,4,2,5,9,2]

largest = second = a[0]

for i in a:
  if i > largest:
    second = largest
    largest = i
  elif i>second and i != largest:
    second = i

print(f"Second highest: {second}")

```


9.Write a Python program to merge two lists but avoid duplicates.

Example:
a = [1,2,3]
b = [3,4,5]
Output → [1,2,3,4,5]

```python
a = [1,2,3]
b = [3,4,5]
c = []

for i in a:
  if i not in c:
    c.append(i)

for i in b:
  if i not in c:
    c.append(i)

print(c)
```
